```python
DataUrl = 'https://raw.githubusercontent.com/Datamanim/pandas/main/AB_NYC_2019.csv'
df = pd.read_csv(DataUrl)
df.head()
```

```python
df.groupby('host_name').size().sort_values(ascending=False).head()
```

```python
ans = df.groupby('host_name', as_index=False).size().rename(columns={'size':'counts'}).sort_values('counts', ascending=False)
ans
```

```python
ans = df.groupby(['neighbourhood_group','neighbourhood'], as_index=False).size()
ans
```

```python
Ans = df.groupby(['neighbourhood_group','neighbourhood'], as_index=False).size().groupby(['neighbourhood_group'], as_index=False).max()
Ans
```

```python
Ans = df[['neighbourhood_group', 'price']].groupby('neighbourhood_group').agg(['mean', 'var', 'max', 'min'])
Ans
```

```python
Ans = df[['neighbourhood_group', 'reviews_per_month']].dropna().groupby('neighbourhood_group').agg(['mean', 'var', 'max', 'min'])
Ans
```

```python
Ans = df.groupby(['neighbourhood_group', 'neighbourhood']).agg({'price':'mean'})
Ans
```

```python
Ans = df.groupby(['neighbourhood_group', 'neighbourhood']).agg({'price':'mean'}).unstack()
Ans
```

```python
Ans = df.groupby(['neighbourhood_group', 'neighbourhood']).agg({'price':'mean'}).unstack().fillna(-999)
Ans
```

```python
Ans = df.loc[df['neighbourhood_group'] == 'Queens'].groupby('neighbourhood').agg({'price':['mean','var','max','min']})
Ans
```

```python
Ans = df.groupby(['neighbourhood_group', 'room_type']).size().unstack().div(df.groupby(['neighbourhood_group']).size(), axis=0)
Ans
```

```python
Ans = pd.crosstab(df['neighbourhood_group'], df['room_type'], normalize='index')
```
